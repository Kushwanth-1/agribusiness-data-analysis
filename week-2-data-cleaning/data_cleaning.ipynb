```python
# Week 2: Agribusiness Data Cleaning
# Internship Project - YuvaIntern

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Agribusiness Data Cleaning Project")
print("Libraries imported successfully!")
```

```python
# Creating a sample agricultural dataset

data = {
    "Date": [
        "28-10-2026",
        "28-10-2026",
        "29-10-2026",
        "2026/10/29",
        "30-10-2026",
        "30-10-2026"
    ],
    "Crop": [
        "Wheat",
        "Wheat",
        "Soybean",
        "soybean",
        "Wheat",
        "Soybean"
    ],
    "Market": [
        "Pune",
        "Pune",
        "Indore",
        "Indore",
        "Nashik",
        "Bhopal"
    ],
    "Price_per_Quintal": [
        2600,
        2600,
        None,
        4200,
        25000,
        4100
    ],
    "Arrivals_in_Quintals": [
        120,
        120,
        85,
        90,
        150,
        110
    ]
}

df = pd.DataFrame(data)

print("Sample Agricultural Dataset:")
display(df)
```

```python
# Checking dataset information

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
# Creating a copy of the original dataset
clean_df = df.copy()

# 1. Remove duplicate records
clean_df = clean_df.drop_duplicates()

# 2. Standardize crop names
clean_df["Crop"] = clean_df["Crop"].str.strip().str.title()

# 3. Standardize market names
clean_df["Market"] = clean_df["Market"].str.strip().str.title()

# 4. Convert dates into a consistent format
clean_df["Date"] = pd.to_datetime(
    clean_df["Date"],
    dayfirst=True,
    format="mixed"
).dt.strftime("%Y-%m-%d")

# 5. Handle missing prices using the median
median_price = clean_df["Price_per_Quintal"].median()

clean_df["Price_per_Quintal"] = clean_df[
    "Price_per_Quintal"
].fillna(median_price)

# Display cleaned dataset
print("Cleaned Agricultural Dataset:")
display(clean_df)
```

```python
# Validating the cleaned dataset

print("Missing Values:")
print(clean_df.isnull().sum())

print("\nDuplicate Records:")
print(clean_df.duplicated().sum())

print("\nData Types:")
print(clean_df.dtypes)

print("\nFinal Statistical Summary:")
display(clean_df.describe())
```

```python
# Export cleaned dataset to CSV

clean_df.to_csv("cleaned_agri_data.csv", index=False)

print("Cleaned dataset exported successfully!")
```
